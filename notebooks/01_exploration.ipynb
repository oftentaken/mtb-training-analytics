```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
df = pd.read_csv("../data/processed/mtb_rides_clean.csv", parse_dates=["start_time", "date"])
df.head()
```

```python
yearly = df.groupby("year").agg(
    rides=("distance_km", "count"),
    total_km=("distance_km", "sum"),
    avg_hr=("avg_heart_rate", "mean")
).round(1)
yearly
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

yearly["rides"].plot(kind="bar", ax=axes[0], color="#2E86AB")
axes[0].set_title("MTB rides per year")
axes[0].set_ylabel("Number of rides")

yearly["total_km"].plot(kind="bar", ax=axes[1], color="#A23B72")
axes[1].set_title("Total distance per year")
axes[1].set_ylabel("Distance (km)")

plt.tight_layout()
plt.savefig("../powerbi/yearly_progress.png", dpi=150)
plt.show()
```
